```python
df["Sales_Representative"] = df["Sales_Representative"].fillna("Unknown")
```

```python
df["Region"] = df["Region"].fillna("Unknown")
```

```python
df["Customer_Type"] = df["Customer_Type"].fillna("Unknown")
```

```python
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
```

```python
data = df.groupby("Product_Category")[["Sales_Amount"]].sum()
```

```python
print(data)
```

```python
pd.pivot_table(df, index = "Region", values = "Quantity_Sold", aggfunc = "sum")
```

```python
sns.countplot(data=df, x="Sales_Representative")
plt.title("Number of Sales by Sales Representative")
plt.xlabel("Sales Representative")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(
    data=df,
    x="Quantity_Sold",
    y="Region",
)
plt.title("Quantity Sold by Region")
plt.xlabel("Quantity Sold")
plt.ylabel("Region")
plt.show()
```

```python
sns.histplot(data=df, x="Unit_Cost", bins=15, kde=True)
plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Quantity_Sold",
    y="Sales_Amount",
    hue="Product_Category",
    alpha=0.6
)
plt.title("Quantity Sold vs Sales Amount")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales Amount")
plt.show()
```

```python
data = df.groupby("Product_Category")["Sales_Amount"].sum()
plt.pie(data, labels=data.index, autopct="%1.1f%%")
plt.title("Sales by Product Category")
plt.show()
```

```python
sns.heatmap(data, annot=True,fmt=".0f")
plt.title("Sales by Product Category")
plt.show()
```
